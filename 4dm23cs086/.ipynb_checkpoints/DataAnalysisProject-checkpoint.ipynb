```python
import pandas as pd
df=pd.read_csv('customer_shopping_behavior.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe() # summary of statistics of numericals
```

```python
df.describe(include="all") #summary staticts of all numericala and catergories
```

```python
#finding nul values
df.isnull()
```

```python
df.isnull().sum()
```

```python
#now the null values is filled by mean and meadian() why mean and meadin(robust outliers)
df['Review Rating']=df.groupby('Category')['Review Rating'].transform(lambda x:x.fillna(x.median()))
```

```python
#Category based we filled the null values
#ex cothing reviw missing then it is replaced by the median of th eclothing only not other
#same as footware review is missing then is it replace by the median of the footware
df.isnull().sum()
```

```python
#snake cases means :everything is lowercase and underscore it is benifial for sql an dpython 
df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(' ','_')
```

```python
df.columns
```

```python
df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
#create a column age_group
labels=['Young_Adult','Adult','Middle-aged','Senior']
df['age_group']=pd.qcut(df['age'],q=4,labels=labels)
```

```python
df[['age','age_group']].head(10)
```

```python
#another_feature :create column purchasr_frequency_days
#purchases is in text form it is difficult to analyse so convert them into numerical
frequency_mapping={
    'Fortnightly':14,
    'Weekly':7,
    'Monthly':30,
    'Quarterly':90,
    'Bi-Weekly':14,
    'Annually':365,
    'Every 3 Months':90
}
df['purchase_frequency_days']=df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['purchase_frequency_days','frequency_of_purchases']].head(10)
```

```python
df[['discount_applied','promo_code_used']].head(10)
```
