```python
#Data manipulation and Data analysis
#first we need to read the data 
import pandas as pd
df = pd.read_csv(r"C:\Users\niree\Downloads\DataAnalysis\data (1).csv")
print(df.head(5))
print(df.tail(5))
print(df.describe())
print(df.dtypes)
```

```python
#df.isnull().any(axis=1)
#here we can see exactly which row contain null values
df.isnull().any()# column wise
```

```python
df.fillna(0) #replacing missing values with 0 
```

```python
#Filling missing values with the mean of the column 
df['Sales_fillna']=df['Sales'].fillna(df['Sales'].mean())
#df.columns
```

```python

df['Sales_fillna']=df['Sales'].fillna(df['Sales'].mean())
df.drop('sales_fillna',axis=1,inplace=True)
df
```

```python
df.columns
```

```python
#Renameing the column
df.rename(columns={'Date':'sales_date'},inplace=True)  #dict used here
df.head()
```

```python
#Changimg datatypess
#it will not work bcz it contains missing values(NaN values)
#df['Value'].fillna(0) # it will not change the missing values permanently it just creating the series 
#df['value_new']=df['Value'].astype(int) 
df['Value'] = df['Value'].fillna(0)  #it will chnage missing values permannetly
df['Value']
```

```python
df['value_new']=df['Value'].astype(int)  #changing the daat typesss
df
```

```python
#we can also apply the function here
df['new values']=df['Value'].apply(lambda x:x*2)
```

```python
#Data aggregating and Grouping
grouped_mean=df.groupby('Product')['Value'].mean()
print(grouped_mean)
```

```python
grouped_sum=df.groupby(['Product','Region'])['Value'].sum()
print(grouped_sum) #sum is add the value in each group
#grouped_sum=df.groupby(['Product','Region']) is creates group object  
```

```python
group_agg=df.groupby('Region')['Value'].agg(['mean','sum','count'])
print(group_agg)
```

```python
######## MERGING  && JOINING##########
```

```python
# sample dataframe

df1=pd.DataFrame({'Key':['A','B','C'],'Value':[1,2,3]})
df2=pd.DataFrame({'Key':['A','B','D'],'Value':[4,5,6]})
#merging here
pd.merge(df1,df2,on="Key",how="inner")# which column name and how the join happen(inner)
#it showing only the samw df keys
```

```python
print(pd.merge(df1,df2,on="Key",how="inner")) # it considering only the same keyy present in the sample
print()
print(pd.merge(df1,df2,on="Key",how="outer")) # it considering each and every row and key .if no values id matched then it will be returned as NaN
print()
print(pd.merge(df1,df2,on="Key",how="left")) #it given more priority  to left side with respect to all the keys taht are available
print()
print(pd.merge(df1,df2,on="Key",how="right")) #it given more priority to the right side with respect to all the keays taht are available 
```

```python
##########   DATA VISUALIZATION ############
#matplotlib
#pip install matplotlib
import matplotlib.pyplot as plt
x=[1,2,3,4,5]
y=[1,4,9,3,7]
#plt.plot(x,y)
plt.plot(x,y,color="red",linestyle="-.")
plt.grid(True)
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.title("Basic Line Plot")
```

```python
x=[1,2,3,4,5]
y1=[1,4,9,16,25]
y2=[1,2,3,4,5]
plt.figure(figsize=(9,5))
plt.subplot(1,2,1)
plt.plot(x,y1,color='green')
plt.title('Plot 1')

plt.subplot(1,2,2)
plt.plot(y1,x,color='red')
plt.title('Plot 2')

```

```python
# BAR PLOT
categories=['A','B','C','D','E']
values=[5,7,3,8,6]
#create a bar plot
plt.bar(categories,values,color='purple')
plt.xlabel('categories')
plt.ylabel('values')
plt.title('Bar plot')
plt.show()
```
