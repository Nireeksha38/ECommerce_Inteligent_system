```python
num = int(input("Enter the number: "))
temp = num
reverse = 0

while temp > 0:
    digit = temp % 10
    reverse = reverse * 10 + digit
    num = temp // 10

if reverse == num:
    print("Palindrome")
else:
    print("Not a palindrome")
```

```python
def fact(n):
    if n==0 or n==1:
        return 1
    else:
        return n*fact(n-1)
n=int(input("enter the elemnt to find fact"))
for i in range(0,n+1):
    print(fact(i))
```

```python

def fibo(n):
    if n==1 or n==0:
        return n
    else:
        return fibo(n-1)+fibo(n-2)
for i in range(0,n+1):
    print(fibo(i))
```

```python
#bubblesort
def bubble_sort(arr):
    n=len(arr)
    for passes in range(0,n):
        for j in range(0,n-1-passes):
            if arr[j]>arr[j+1]:
                arr[j],arr[j+1]=arr[j+1],arr[j]
    return arr


arr=[5,4,3,2,1]
sorted_list=bubble_sort(arr)
print("sorted element",sorted_list)
```
