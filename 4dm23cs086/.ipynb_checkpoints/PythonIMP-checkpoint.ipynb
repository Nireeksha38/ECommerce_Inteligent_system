```python
#bubblesort
def bubble_sort(arr):
    n=len(arr)
    for passes in range(0,n):
        for j in range(0,n-1-passes):
            if arr[j]>arr[j+1]:
                arr[j],arr[j+1]=arr[j+1],arr[j]
    return arr
arr=[23,54,12,63,23,43,56]
sorted_list=bubble_sort(arr)
print("Sorted_list:",sorted_list)
```

```python
#bubble sort
def bubble_sort(arr):
    n=len(arr)
    for passes in range(0,n):
        for j in range(0,n-1-passes):
            if arr[j]>arr[j+1]:
                arr[j],arr[j+1]=arr[j+1],arr[j]
    return arr
arr=[23,45,12,6,1,2,34,56]
sorted_list=bubble_sort(arr)
print("Sorted list:",sorted_list)
```

```python
#selection sort
def selection_sort(arr):
    n=len(arr)
   
    for i in range(n-1):
       mid_index=i
       for j in range(i+1,n):
           if arr[j]<arr[mid_index]:
               mid_index=j
       arr[i],arr[mid_index]=arr[mid_index],arr[i]

    return arr
arr=[12,25,11,34,90,22]
sorted_list=selection_sort(arr)
print(sorted_list)
```

```python
def selection_sort(arr):
    n=len(arr)
    for i in range(n-1):
        min_index=i
        for j in range(i+1,n):
            if arr[j]<arr[min_index]:
                min_index=j
        arr[i],arr[min_index]=arr[min_index],arr[i]
    return arr

arr=[12,34,52,12,34,67]
sorted_list=selection_sort(arr)
print(sorted_list)
```

```python
#python inheritance
class Car:
    def __init__(self,windows,doors):
        self.windows=windows
        self.doors=doors
    def info(self):
        print(f" The car having {self.doors} amd {self.windows}")
class Tesla(Car):
    def __init__(self,window,doors,enginetype):
        super().__init__(window,doors)
        self.enginetype=enginetype
    def info(self):
        print(f" The car having {self.doors} amd {self.windows} and the enginetype is {self.enginetype}")

car=Car(4,2)
car.info()
car1=Tesla(4,2,"petrol")
car1.info()
```

```python
def linear_search(arr,target):
    size=len(arr)-1
    for i in range(0,size):
        if arr[i]==target:
            return i
            break
    else:
        return -1


arr=[1,6,7,35]
target=6
result=linear_search(arr,target)
```

```python
#Two sum problem
def two_sum(arr,target):
    for i in range(len(arr)):
        for j in range(i+1,len(arr)):
            if arr[i]+arr[j]==target:
                return [i,j]
arr=[6,7,3,7,9,2]
target=10
result=two_sum(arr,target)
print(result)
```

```python
#Three sum
def three_sum(arr,target):
    for i in range(len(arr)):
        for j in range(i+1,len(arr)):
            for k in range(i+2,len(arr)):
                if arr[i]+arr[j]+arr[k]==target:
                    return [i,j,k]
arr=[1,6,7,3,7,9,2]
target=10
result=three_sum(arr,target)
print(result)
```

```python
#print fibonacci series u to n
n=int(input("enter a number to get a series"))
def fib(n):
    if n==0 or n==1:
        return n
    else:
        return fib(n-1)+fib(n-2)
for i in range(0,n+1):
    print(fib(i),end=" ")
```

```python
#find a factorial of a number
def fact(n):
    if n==0 or n==1:
        return 1
    else:
        return n*fact(n-1)
fact(5)
```
