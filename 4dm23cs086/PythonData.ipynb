```python
##Inheritance
class Car:
    def __init__(self,windows,doors,enginetype):
        self.windows=windows
        self.doors=doors
        self.enginetype=enginetype
    def drive(self):
        print(f" The car is having {self.windows} windows and {self.doors} doors and the enginetype is {self.enginetype} ")

car=Car(4,2,"petrol")
car.drive()

class Tesla(Car):
    def __init__(self,windows,doors,enginetype,isselfDrivingcar):
        super().__init__(windows,doors,enginetype)
        self.isselfDrivingcar=isselfDrivingcar
    def drive(self):
        print(f"The car is driving using {self.enginetype} and is is also a self driving car .{self.isselfDrivingcar}")

tesla1=Tesla(4,2,"petrol",True)
tesla1.drive()
```

```python
num = int(input("Enter the number: "))
temp = num
reverse = 0

while temp > 0:
    digit = temp % 10
    reverse = reverse * 10 + digit
    num = temp // 10

if reverse == num:
    print("Palindrome")
else:
    print("Not a palindrome")
```
