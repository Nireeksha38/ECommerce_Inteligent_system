```python
#bubblesort
def bubble_sort(arr):
    n=len(arr)
    for passes in range(0,n):
        for j in range(0,n-1-passes):
            if arr[j]>arr[j+1]:
                arr[j],arr[j+1]=arr[j+1],arr[j]
    return arr


arr=[5,5,3,0,1,85]
sorted_list=bubble_sort(arr)
print("sorted element",sorted_list)
```
